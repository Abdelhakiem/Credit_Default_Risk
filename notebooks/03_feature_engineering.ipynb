```python
from pathlib import Path

import numpy as np
import pandas as pd
import re
import sys
from IPython.display import Markdown, display
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

INTERIM_DIR = Path("../data/interim")

train_df = pd.read_parquet(INTERIM_DIR / "train.parquet")
test_df = pd.read_parquet(INTERIM_DIR / "test.parquet")

print("Train:", train_df.shape)
print("Test: ", test_df.shape)
display(train_df.head())
```

### Functions

```python
def safe_divide(numerator, denominator):
    denominator = denominator.where(denominator > 0)
    return numerator.div(denominator).replace([np.inf, -np.inf], np.nan)
```

```python
def add_time_features(df):
    df = df.copy()

    employed_days = df["DAYS_EMPLOYED"].replace(365243, np.nan)
    df["DAYS_EMPLOYED"] = employed_days

    df["AGE_YEARS"] = -df["DAYS_BIRTH"] / 365.25
    df["EMPLOYMENT_YEARS"] = -employed_days / 365.25

    df["AGE_BIN"] = pd.cut(
        df["AGE_YEARS"],
        bins=[0, 25, 35, 45, 55, 65, np.inf],
        labels=["<25", "25-34", "35-44", "45-54", "55-64", "65+"],
        right=False,
    )

    df["EMPLOYMENT_BIN"] = pd.cut(
        df["EMPLOYMENT_YEARS"],
        bins=[0, 1, 3, 5, 10, 20, np.inf],
        labels=["<1", "1-2", "3-4", "5-9", "10-19", "20+"],
        right=False,
    )

    return df
```

```python
def add_application_ratios(df):
    df = df.copy()

    df["CREDIT_TO_INCOME_RATIO"] = safe_divide(
        df["AMT_CREDIT"], df["AMT_INCOME_TOTAL"]
    )
    df["ANNUITY_TO_INCOME_RATIO"] = safe_divide(
        df["AMT_ANNUITY"], df["AMT_INCOME_TOTAL"]
    )
    df["GOODS_TO_CREDIT_RATIO"] = safe_divide(
        df["AMT_GOODS_PRICE"], df["AMT_CREDIT"]
    )
    df["CREDIT_TO_ANNUITY_RATIO"] = safe_divide(
        df["AMT_CREDIT"], df["AMT_ANNUITY"]
    )
    df["INCOME_PER_FAMILY_MEMBER"] = safe_divide(
        df["AMT_INCOME_TOTAL"], df["CNT_FAM_MEMBERS"]
    )

    return df
```

```python
def select_property_suffix(df, suffix="MEDI"):
    df = df.copy()
    property_suffixes = ("AVG", "MEDI", "MODE")

    families = {
        col.rsplit("_", 1)[0]
        for col in df.columns
        if col.endswith(tuple(f"_{s}" for s in property_suffixes))
        and all(
            f"{col.rsplit('_', 1)[0]}_{s}" in df.columns
            for s in property_suffixes
        )
    }

    drop_cols = [
        f"{family}_{s}"
        for family in families
        for s in property_suffixes
        if s != suffix
    ]

    return df.drop(columns=drop_cols)
```

```python
def engineer_features(df, property_suffix="MEDI"):
    result = add_time_features(df)
    result = add_application_ratios(result)
    result = select_property_suffix(result, suffix=property_suffix)

    return result
```

### 1. Clean employment and create age/employment features

```python
train_step = add_time_features(train_df)

display(
    train_step[
        ["DAYS_EMPLOYED", "EMPLOYMENT_YEARS", "EMPLOYMENT_BIN",
         "DAYS_BIRTH", "AGE_YEARS", "AGE_BIN"]
    ].head()
)

print("Employment missing after cleaning:", train_step["DAYS_EMPLOYED"].isna().sum())
```

### 2. Application amount ratios

```python
train_step = add_application_ratios(train_step)

ratio_cols = [
    "CREDIT_TO_INCOME_RATIO",
    "ANNUITY_TO_INCOME_RATIO",
    "GOODS_TO_CREDIT_RATIO",
    "CREDIT_TO_ANNUITY_RATIO",
    "INCOME_PER_FAMILY_MEMBER",
]

display(train_step[ratio_cols].describe(percentiles=[0.01, 0.5, 0.99]).T)
```

### 3. Select Properity Suffix

```python
# selecting property suffix MEDI for the train_step dataframe : highest feature importance
train_step = select_property_suffix(train_step, suffix="MEDI")

print("Property AVG columns remaining:",
      [c for c in train_step if c.endswith("_AVG")])
print("Property MEDI columns:",
      [c for c in train_step if c.endswith("_MEDI")])
```

### 4. Full feature engineering pipeline:

```python
train_features = engineer_features(train_df)
test_features = engineer_features(test_df)

assert train_features.columns.tolist() == test_features.columns.tolist()
assert len(train_features) == len(train_df)
assert len(test_features) == len(test_df)

print("Train:", train_features.shape)
print("Test: ", test_features.shape)
```
